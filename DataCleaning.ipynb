# Outlier

```python
df.columns
```

```python
print("Dataset Shape:",df.shape)
print("\n Marks Stastics before oulier treatment:")
print(df["Salary"].describe)
```

```python
q1=df["Salary	"].quantile(0.25)
q3=df["Salary	"].quantile(0.75)

IQR=q3-q1

lower_limit=q1-1.5 * IQR
upper_limit=q3+1.5 * IQR

print("Q1: ",q1)
print("Q3: ",q3)
print("IQR: ",IQR)
print("Lower Limit: ",lower_limit)
print("Upper Limit: ",upper_limit)
outlier= df[(df["Salary	"]<lower_limit)|
            (df["Salary	"]>upper_limit)
            ]
print(len(outlier))
print(outlier[["ID","Salary	"]].head(20))
```

```python
import matplotlib.pyplot as plt
plt.figure(figsize=(8,5))
plt.boxplot(df["Salary	"])
plt.title("Salary Before Outlier")
plt.ylabel("Salary")
plt.show()
```

```python
import matplotlib.pyplot as plt
import pandas as pd
df=pd.read_excel(r"C:\Users\SUJAN SURE\OneDrive\Desktop\AIML-3\virtual\Scripts\data.xlsx")

avg_salary = df.groupby('JobCity')['Salary'].mean()

positions = range(len(avg_salary))

plt.figure(figsize=(6,6))
plt.bar(positions, avg_salary.values, color=['green','yellow'])

plt.xticks(positions, avg_salary.index, rotation=45)

plt.title("Average Salary by JobCity")
plt.xlabel("JobCity")
plt.ylabel("Average Salary")

plt.tight_layout()
plt.show()
```
