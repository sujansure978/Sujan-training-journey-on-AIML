Loops

```python
data=[15,10,20,14,30,25]
new=[]
for i in data:
    result=i*2
    new.append(result)
print(new)
for i in range(5,15):
    print(i)
for i in range(5,15):
    print(i,end='\t')
for i in range(5,15,2):
    print(i)
```

Statistics

```python
import statistics
data=[1,2,3,4,5]
print(statistics.mean(data))
print(statistics.median(data))
print(statistics.mode(data))
print(statistics.pvariance(data))
print(statistics.pstdev(data))
```

```python
import numpy as np 
data=[1,2,3,4,5]
print(np.mean(data))
print(np.median(data))
print(np.var(data))
print(np.std(data))
```
