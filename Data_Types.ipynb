# with open

```python
#with open "a-->append",r-->read,w-->write
with open("Filename100.pdf",mode="w") as f:
    f.write("Hello World!!")
    f.write("\n")
    f.write("Welcome to  our class python,\n")
```

```python
#with open "a-->append",r-->read,w-->write
with open("Filename100.pdf",mode="a") as f:
    f.write("hello how are you?")
    f.write("\n")
    f.write("I am fine with learning python\n")
```

```python
#with open "a-->append",r-->read,w-->write
with open("Filename100.pdf",mode="r") as f:
    data=f.read()
    print(data)
```
