```python
import pandas
import seaborn as sns
import matplotlib.pyplot as plt
df=pd.read_csv(r"C:\Users\SUJAN SURE\OneDrive\Desktop\AIML-3\virtual\Scripts\Cyber_salaries.csv")
print(df.head())
numeric_id=df.select_dtypes(include=["number"])
corr=numeric_id.corr()
plt.figure(figsize=(10,8))
sns.heatmap(corr,annot=True,cmap="coolwarm",fmt=".2f")
plt.title("Correlation Heatmap of Numeric Features in diamonds Dataset")
plt.show()
```

```python
print("Dataset Shape:",df.shape)
print("\n Salary statistics before outlier treatment:")
print(df['salary'].describe())
q1 = df['salary'].quantile(0.25)
q3 = df['salary'].quantile(0.75)

IQR = q3-q1

lower_limit = q1 - 1.5 * IQR
upper_limit = q3 + 1.5 * IQR

print("Q1:",q1)
print("Q3:",q3)
print("IQR:",IQR)
print("Lower:",lower_limit)
print("Upper:",upper_limit)

outliers = df[
    (df["salary"] < lower_limit) |
    (df["salary"] > upper_limit)
]

print("\nNumber of Salary Outliers:", len(outliers))

print("\nsalary Outliers:")
print(outliers[["work_year", "salary"]].head(20))
```
