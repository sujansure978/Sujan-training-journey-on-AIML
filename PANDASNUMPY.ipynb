```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm
x=np.linspace(-4,4,100)
y=norm.pdf(x)
plt.plot(x,y)
plt.title("Normal Distribution")
plt.xlabel("X values")
plt.ylabel("Probability Density")
plt.show()
```

Over View Of Correlation

```python
import numpy as np
marks = np.array([45, 67, 78, 92, 85, 49, 95, 60, 80, 38, 91])
between_60_80 = np.sum((marks >= 60) & (marks <= 80))
above_90 = np.sum(marks > 90)
below_50 = np.sum(marks < 50)
print("Students scoring between 60 and 80:", between_60_80)
print("Students scoring above 90:", above_90)
print("Students scoring below 50:", below_50)
```

```python
import pandas
import seaborn as sns
import matplotlib.pyplot as plt
df=sns.load_dataset("diamonds")
print(df.head())
numeric_id=df.select_dtypes(include=["number"])
corr=numeric_id.corr()
plt.figure(figsize=(10,8))
sns.heatmap(corr,annot=True,cmap="coolwarm",fmt=".2f")
plt.title("Correlation Heatmap of Numeric Features in diamonds Dataset")
plt.show()
```
